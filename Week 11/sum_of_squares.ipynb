Can we write $2x^2-2x+1$ as a sum of squares? If we can, the identity proves the polynomial is nonnegative.

```python
from sympy import symbols
from SumOfSquares import SOSProblem

x = symbols("x")
p = 2*x**2 - 2*x + 1

sos = SOSProblem()
constraint = sos.add_sos_constraint(p, [x])
sos.solve(solver="cvxopt")
sum(constraint.get_sos_decomp())
```

The solver uses a Gram matrix internally. `get_sos_decomp()` gives us the square terms with numerical coefficients. [SumOfSquares.py example](https://github.com/yuanchenyang/SumOfSquares.py#examples)

For `external_sos`, Python recovers exact coefficients and Lean checks the weights and polynomial identity. [find_sos.py](find_sos.py)
